```python
# input the raw sequence
raw_seq = """ATGAATCAACATAATACCCAAATCAATAAATTTATCTTTTTTAGTAGCTTCAGAATCATCAGTACCACCCCAATTCAATTTATCAACAATAATAGGACCAGCAACAGCAATAGCAACATATTCATCATTAGAAGCAGGAGAATTCAAAACAATACCAACAGCTTTTTTATTATCAGCAGTATTCCAAGTTTTACAAACAGCACCATTAGAATTACCTTTTTCACAATCACCAGCATGCAAATTATCAATACCAGAATTTTTATGAGATCTAATATGAACCAAACCCAT"""

# load the sequence to Bio
seq = Seq("".join(raw_seq.split()).upper())

print(f"Sequence length: {len(seq)} nucleotides")
```

```python
# GC and AT content
gc = gc_fraction(seq)
at = 1 - gc

print(f"GC content: {gc:.2%}")
print(f"AT content: {at:.2%}")
```
